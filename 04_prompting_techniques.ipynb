## Part 1: Prompting Fundamentals

### 1.1 Zero-Shot Prompting

```python
# Example prompts for digital twin
zero_shot_examples = [
    {
        'task': 'activity_prediction',
        'prompt': 'Predict the next activity: I just finished work, had dinner, and now...',
        'expected': 'leisure/relaxation activity'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences from: I love outdoor activities but hate crowded places',
        'expected': {'likes': ['outdoor', 'quiet'], 'dislikes': ['crowds']}
    }
]

print('Zero-Shot Examples:')
for ex in zero_shot_examples:
    print(f"\nTask: {ex['task']}")
    print(f"Prompt: {ex['prompt']}")
    print(f"Expected: {ex['expected']}")
```

**Task**: Design prompts for your digital twin use case.

```python
my_prompts = [
    # 1. Next-activity prediction from current state and routine
    (
        "You are a digital twin modeling a busy weekday schedule with "
        "morning work/teaching blocks, afternoon prep, and evening rest. "
        "Given current time = 17:30, energy = 0.65 (0-1 scale), and recent "
        "activity = 'work', predict the most likely next activity, give a "
        "rough probability, and briefly explain why."
    ),
    # 2. What-if scenario simulation
    (
        "Act as a personal routine digital twin. Compare the likely evening "
        "activity under two scenarios when current activity is 'study/coding': "
        "(A) sleep last night was 7.5 hours and energy is high, vs. "
        "(B) sleep last night was 4.5 hours and energy is low. Highlight how "
        "the probability shifts."
    ),
    # 3. Sequence / multi-step rollout
    (
        "Using a 14-day routine pattern where 'work' (code 1) occurs ~71% of "
        "days, rest/off-day (code 0) ~14%, and occasional special activities "
        "(codes 2 and 4) ~7% each, simulate a plausible 5-step sequence starting "
        "from 'work' and note where probabilistic branching could happen."
    ),
    # 4. Constraint-aware decision support
    (
        "You are a digital twin helping optimize daily habits. Today has only "
        "90 minutes of open evening time, moderate energy (0.5/1.0), and an "
        "upcoming morning commitment. Recommend the highest-fit activity and "
        "explain the trade-off against leisure or deep focus."
    ),
    # 5. Structured JSON output for programmatic use
    (
        "Act as a digital twin predictor for state "
        "{'day': 'Saturday', 'hour': 15, 'weather': 0.25, 'mood': 0.65}. "
        "Respond ONLY with valid JSON containing keys: "
        "'predicted_activity', 'probability' (0.0-1.0), "
        "'key_factors', and 'fallback_activity'."
    ),
]

for i, prompt in enumerate(my_prompts, start=1):
  print(f"--- Prompt {i} ---\n{prompt}\n")
```
